```python
import pandas as pd
```

```python
dd= pd.read_csv('../dataset/LMS_Dirty_Dataset.csv')
dd
```

```python
dd.head(10) #first 10 data 
dd.tail(10) #last 10 data ko
dd.shape #350 ota rows and 18 ota column raixa
dd.columns
dd.dtypes 
dd.describe() #numerical value ko matra
```

```python
dd[dd.duplicated()].sum()
#duplicate data xaina so hamile remove garna parena
dd.info()
dd.isnull().sum() #euta only duplicate xa ie attendence_percent only
dd['attendance_percent']=dd['attendance_percent'].fillna(dd['attendance_percent'].mean())
dd.isnull().sum()#now nothing is null or NaN
```

```python
#incorrect datatypes
dd['course_fee']=pd.to_numeric(dd["course_fee"], errors="coerce")
dd['age']=pd.to_numeric(dd["age"], errors="coerce")
dd['discount']=pd.to_numeric(dd["discount"], errors="coerce")
dd['assignment_score']=pd.to_numeric(dd["assignment_score"], errors="coerce")
dd['final_score']=pd.to_numeric(dd["final_score"], errors="coerce")
dd['rating']=pd.to_numeric(dd["rating"], errors="coerce")
dd['attendance_percent']=pd.to_numeric(dd["attendance_percent"], errors="coerce")
```

```python
#incosistent text value
dd['city'].unique()
dd["city"] = dd["city"].replace("ktm", "Kathmandu")
dd['course'].unique()
dd['course']=dd['course'].replace('Pythn AI','Python with AI/ML')
dd['gender'].unique()# its ok
dd['payment_status'].unique()
dd['completion_status'].unique()
dd["city"] = dd["city"].str.strip().str.title()
```

```python
#incorrect datatypes 
dd[(dd["age"] < 0)|(dd["age"] >100)]#--> correct xa
dd[dd['course_fee']<0] # 1 data xa 
dd['course_fee']=dd['course_fee'].abs() # kinaki 5000 xa aru data ko value fluuter ma
# dd[dd['course_fee']<0] now xaina
dd[dd['discount']<0]

dd[dd['attendance_percent']<0]
dd[(dd['assignment_score']<0) |(dd['assignment_score']>100)] #1 value in id 39
dd['assignment_score']=dd['assignment_score'].abs()
dd[dd['final_score']<0]#its ok
dd[(dd['rating']<0)|(dd['rating']>5)]
```

```python
# age
Q1=dd['age'].quantile(0.25)
Q3=dd['age'].quantile(0.75)
IQR = Q3-Q1
lower_bound_age=Q1-1.5*IQR
upper_bound_age= Q3+ 1.5 *IQR
outlier = dd[(dd['age']<lower_bound_age)| (dd['age']>upper_bound_age)]
outlier
```

```python
#course_fee
Q1=dd['course_fee'].quantile(0.25)
Q3=dd['course_fee'].quantile(0.75)
IQR = Q3-Q1
lower_bound_cour=Q1-1.5*IQR
upper_bound_cour= Q3+ 1.5 *IQR
outlier = dd[(dd['course_fee']<lower_bound_cour)| (dd['course_fee']>upper_bound_cour)]
outlier # values 1
```

```python
#discount
Q1=dd['discount'].quantile(0.25)
Q3=dd['discount'].quantile(0.75)
IQR = Q3-Q1
lower_bound_dis=Q1-1.5*IQR
upper_bound_dis= Q3+ 1.5 *IQR
outlier = dd[(dd['discount']<lower_bound_dis)| (dd['discount']>upper_bound_dis)]
outlier # no any values
```

```python
#attendance_percent
Q1=dd['attendance_percent'].quantile(0.25)
Q3=dd['attendance_percent'].quantile(0.75)
IQR = Q3-Q1
lower_bound_att=Q1-1.5*IQR
upper_bound_att= Q3+ 1.5 *IQR
outlier = dd[(dd['attendance_percent']<lower_bound_att)| (dd['attendance_percent']>upper_bound_att)]
outlier  # no any values
```

```python
## rating and final score donot need because we already tested range of them
```

```python
dd_clean =dd[(dd['course_fee']>lower_bound_cour) &(dd['course_fee']<upper_bound_cour)]
#removed id  19
```

```python
#clip method
dd['course_fee']=dd['course_fee'].clip(upper_bound_cour,lower_bound_cour)
dd.loc[dd['student_id']==9] 
```

```python
Median =dd['course_fee'].median()
dd.loc[(dd['course_fee']<lower_bound_cour) | (dd['course_fee']>upper_bound_cour),'course_fee']=Median
dd.loc[dd['student_id']==9]#checking
dd.loc[dd['student_id']==19]#sabai thik xa
```

```python
# for course fee
Mean=dd['course_fee'].mean()
SD=dd['course_fee'].std()
dd['z_score_course_fee']=(dd['course_fee']-Mean)/SD
dd["z_score_course_fee"]
outlier = dd[(dd['z_score_course_fee']>3)|(dd['z_score_course_fee']<-3)]
outlier #id= 19
```

```python
Mean=dd['attendance_percent'].mean()
SD=dd['attendance_percent'].std()
dd['z_score_att']=(dd['attendance_percent']-Mean)/SD
dd["z_score_att"]
outlier = dd[(dd['z_score_att']>3)|(dd['z_score_att']<-3)]
outlier
```

```python
Mean=dd['assignment_score'].mean()
SD=dd['assignment_score'].std()
dd['z_score_assign']=(dd['assignment_score']-Mean)/SD
dd["z_score_assign"]
outlier = dd[(dd['z_score_assign']>3)|(dd['z_score_assign']<-3)]
outlier #   id 39 ma
```

```python
Mean=dd['final_score'].mean()
SD=dd['final_score'].std()
dd['z_score_fs']=(dd['final_score']-Mean)/SD
dd["z_score_fs"]
outlier = dd[(dd['z_score_fs']>3)|(dd['z_score_fs']<-3)]
outlier
```
