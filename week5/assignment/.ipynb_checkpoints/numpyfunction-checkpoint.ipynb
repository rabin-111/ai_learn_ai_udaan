```python
import numpy as np
```

```python
# 1. numpy sort()
#  purpose->sorts the value in array
# syntax->np.sort (array)
arr=np.array([4,3,5,7,1])
print(np.sort(arr))
```

```python
# 2. np.unique()
# purpose->it finds unique value out of duplicated values in array and gives output in sorted order
#syntax -> np.unique(array)
a=np.array([1,1,1,2,3,4,6,7,8,4])
print(np.unique(a))
```

```python
#3.np.argsort()
# i found it little confusing to understand but its simple
# purpose->it returns the indices value of the original array so that the value gives the array in sorted order
# syntax-> array_name.argsort() or np.argsort(array_name) both runs 
indexx =np.argsort(arr)#indexx=arr.argsort()
print(indexx) # yo value chai orginal array ko index lai sort garxa suru ma
print(arr[indexx]) # using that value sorts the orignal array
```

```python
#4. np.where()
#note use this in sorted array
# purpose-> it helps to use conditional statements. It returns the indices of the original array that satisgy the condition
#np.where(condition)
w=np.array([1,2,3,4,5,6])
res=np.where(w>4)
print(res)
```

```python
#5. np.concatenate()
# purpose -> to combine two arrays around an axis
# syntax np.concat((arrays,.,axis=0),out=None,dtype=None)
a=np.array([1,2,3,4,5])
b=np.array([2,3,5,6,7])
c=np.concatenate((a,b)) # yeha bhitra value put garney() -> if only used (a,b) error 
print(c)

#for 2D array
aa=np.array([[1,2],[3,4]])
bb=np.array([[2,4],[5,6]])
cc=np.concat((aa,bb))
print(cc)
```

```python
#6. np.vstack()
# purpose is to stack the arrays one after another vertically
# syntax same as concat -> np.vstack((arrays,..))
a=np.array([1,2,3])
b=np.array([1,2,3])
c=np.vstack((a,b))
print(c)
# it basically converts 1 D (N,)to (1,N)
# for 2D or higher it acts same as .concat()
a=np.array([[1,2],[2,3]])
b=np.array([[4,5],[6,7]])
c=np.vstack((a,b))
print(c)
```

```python
#7. np.hstack()

#purpose is to merge horizontally thw two arrays
# syntax np.hstack(arrays,..)
# c=np.hstack((a,b))
# print(c)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.hstack((a, b)))

```

```python
# 8.  np.split()
# purpose -> to split or divide an array into equal sub arrays
# syntax np.split(array,how_many_parts)
a=np.array([1,2,3,4,5,6])
print(np.split(a,3))
#note if yesle divide garna sakena bhaney it throws error
```

```python
#9. np.clip()  # we read earlier in winsorization
# purpose is to clip or limit the value  of the array to a certain range 
# syntax -> np.clip(array,a_min=value,a_max=value,out=None)
np.clip(a,a_min=2,a_max=5)
# yedi value less than the limit then it gives the min or max value
```

```python
#10. np.argmax()
# purpose is return the index of largest value in an array
#syntax-> np.argmax(array)
index=np.argmax(a)
print(index)
```

```python
#11. np.argmin()
# purpose _ returns the index of smalles value in an array
#syntax -> np.argmin(array)
index =np.argmin(a)
print(index)
```

```python
#12. np.random.randint()
#purpose is to return a random array
# syntax -> np.random.randint(lower,upper,size=N)
num = np.random.randint(10) # gives one integer from 0-9
print(num)
# for range between 2 to 6
n=np.random.randint(2,6)
print(n)
# to generate 1D array
n1=np.random.randint(2,8,size=3)
print(n1)
```

```python
#13 np.random.choice()
# purpose is to return random value from the array
a=np.array([1,2,3,4,5,6])
n=np.random.choice(a)
print(n)
```

```python
#14 np.transpose()
# purpose is to reverse the axes or in mathematical term tranpose of a matrix
# it doesnot have any effect in 1D array
# it have effect in more than 2D finds the tranpose or rotate axes
a=np.array([[1,2,3],[5,6,7]])
print(np.transpose(a))
```

```python
#15 . np.linspace() 
# purpose is to create a equally spaced numbers in an arrays
# suntax    np.linspace(min,max,num=N)
print(np.linspace(1,5,num=7))
```

```python
# 16 np.zeros()
#is to return the  array with required zeros
# syntax np.zeros(5)
print(np.zeros(5))
```

```python
# 17 np.ones()
# same as 16 but with 1
# syntax np.ones(4)
print(np.ones(5))
```

```python
# 18. np.eye(value) is to create the identity matrix ie 1 in diagonal and square matrix
print(np.eye(5))
```
