```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#boxplot without distortion
ds1=pd.DataFrame({
    'name':['rabin','nabin','abin','bin','tin','yin'],
    'salary':[1000,2000,3000,4000,5000,6000]
})
plt.boxplot(ds1['salary'])
```

```python
#boxplot with outliers
ds2=pd.DataFrame({
    'name':['rabin','nabin','abin','bin','tin','yin','ok'],
    'salary':[1000,2000,3000,4000,5000,6000,50000]
})
plt.boxplot(ds2['salary'])# outliers can be seen in 50000 and the box is also shrinked
```
