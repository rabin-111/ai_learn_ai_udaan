```python
import numpy as np
```

```python
# vectorization
```

```python
a1=np.array([1,2,3])
res=a1+1
print(res)
a1
```

```python
res2=a1*3
res2
```

```python
res3=a1>1
res3
```

```python
#custom functions
vec=np.vectorize(lambda x:x**2)
res=vec(a1)
print(res)
```

```python
a2=np.array([[1,2],[3,4]])
a3=np.array([[4,5],[6,7]])
res=np.dot(a2,a3)
print(res)
```

```python
r1=a1.sum()
print(r1)
```

```python
#broadcasting
```

```python
res=a1+1 #-->vectorization uses the concept of broadcasting it scales 1 to required shape of a1
```

```python
#broadcasting 1D to 2D
x=[2,3]
res=x+a2 #scales x to 2D
print(res)
```

```python
#broadcasting for matrix multiplication
m=np.array([[1,2,3],
          [3,4,5],
          [5,6,7]])
s=np.array([10,20,30])
res=m*s
print(res)
```

```python
# class
res=s*10
r=(res>10) & (res<50)
print(r)
```

```python
#broadcasting not used when the shape of two array is same
```

```python
#data reshaping
a=np.array([1,2,3,4,5,6])
res=a.reshape(6,1)
res2=a.reshape(2,3)
print(res)
print(res2)
```

```python
arr2=np.array([[1,2,3],
             [4,5,6]])
arr2.shape

res=arr2.reshape(6,1)
print(res)
```

```python
res=arr2.reshape(-1)
print(res) # shortcut for automatically getting the dimension we dont need to calculate
# how does it work?  it works such that i auto calculates the dimension required to get the result! 
```
