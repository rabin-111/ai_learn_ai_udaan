```python
import numpy as np
```

```python
arr=np.array([10,20,30,40])
print(arr+5) #number jastai addition garna sakinxa
```

```python
#accessing the values of 1D arrays
print(arr[0])
print(arr[-1])
print(arr[0:3]) # slicing the 1D array
```

```python
#accessing the values of 2D array
arr2=np.array([
    [21,200],
    [22,300],
    [23,400]
])
print(arr2[2,0])
```

```python
#slicing the 2Darray
arr3=arr2[0:2]
print(arr3)
print(arr2[:,1])# gives all rows but only column 1
```
