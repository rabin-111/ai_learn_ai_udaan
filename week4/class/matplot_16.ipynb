```python
#real life example
dd=pd.read_csv('../../week3/dataset/daraz_dataset.csv')
category=dd.groupby('Category')['Product'].count()
```

```python
plt.figure(figsize=(20,8))# figure
plt.bar(category.index,category.values,color='green',edgecolor='blue')# coloring ko lagi
plt.xlabel('categories',fontsize=30)
plt.ylabel('products',fontsize=30)
plt.title('products and categories bar graph',fontsize=40)
plt.grid(axis='y',linestyle ='--',alpha=0.5)# alpha for opacity line style for line structure and asix for which axis grid is needed
```
