```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dd=pd.read_csv('../../week3/dataset/daraz_dataset.csv')
dd
```

```python
dfd=pd.DataFrame({
    'Age':[21,22,23,24,25,26,27],
    'TotalAmount':[1000,2000,1500,1800,2000,2200,2500]
})
plt.scatter(dfd['Age'],dfd['TotalAmount'],s=80,color="blue",alpha=0.5)
plt.title("relation between age and totalamount")
plt.xlabel("age")
plt.ylabel('totalamount')
plt.colorbar()
```
