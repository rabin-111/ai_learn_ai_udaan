```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dd=pd.read_csv('../../week3/dataset/daraz_dataset.csv')
category=dd.groupby('Category')['Product'].count()
category
```

```python
myexplode=[0.4,0,0,0,0,0,0]
plt.pie(category.values,labels=category.index,explode=myexplode,startangle=90,shadow=True)
#plt.legend(title='category of products')-- description of the data
```

```python
# assignment grouping gender and product
gender=dd.groupby('Gender')['Product'].count()
gender
plt.pie(gender.values,labels=gender.index,colors=['green','blue'])
```

```python
plt.hist(dd['UnitPrice'],color='blue')
plt.xlabel('UnitPrice')
plt.ylabel('Frequency')
plt.title('unit price histogram',fontsize=20)
```

```python
#custom bins
bins=[0,5000,10000,30000,50000,80000]
plt.hist(dd['UnitPrice'],bins=bins,edgecolor='black')
```
